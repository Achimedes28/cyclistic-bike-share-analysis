```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
Trip2019 = pd.read_csv("Trips_2019_Q1.csv")
Trip2020 = pd.read_csv("Trips_2020_Q1.csv")
```

```python
Trip2019.head(5)
```

```python
Trip2020.head(5)
```

```python
"""
Mengambil kolom yang diperlukan dari dataframe Trip2019 dan Trip2020
"""

columns_needed19 = ["trip_id", "start_time","end_time","from_station_name","to_station_name","usertype","ride_lengt","day_of_week","day_name"]
df2019 = Trip2019[columns_needed19].copy()

columns_needed20 = ["ride_id","started_at","ended_at","start_station_name","end_station_name","member_casual","ride_length","day_of_week","day_name"]
df2020 = Trip2020[columns_needed20].copy()
```

```python
df2019.head(5)
```

```python
df2020.head(5)
```

```python
"""
Mengecek apakah ada missing value pada dataframe
"""
df2019.isnull().sum()
```

```python
"""
Mengecek apakah ada missing value pada dataframe
"""
df2020.isnull().sum()
```

```python
"""
Menghapus missing value pada dataframe
"""
df2019.dropna(inplace = True)
df2020.dropna(inplace = True)
```

```python
"""
Mengecek apakah ada missing value pada dataframe setelah dihapus
"""
df2019.isnull().sum()
```

```python
"""
Mengecek apakah ada missing value pada dataframe setelah dihapus
"""
df2020.isnull().sum()
```

```python
"""
Mengubah nama kolom pada dataframe Trip2019
"""
rename_2019 = {
    "trip_id": "ride_id",
    "start_time": "started_at",
    "end_time": "ended_at",
    "from_station_name": "start_station_name",
    "to_station_name": "end_station_name",
    "usertype": "member_casual"}

df2019 = df2019.rename(columns=rename_2019)
```

```python
"""
Mengecek apakah column 2019 sudah sesuai dengan 2020
"""
print(df2019.columns.tolist())
print(df2020.columns.tolist())
```

```python
"""
Mengubah nama kolom pada dataframe Trip2019
"""
df2019 = df2019.rename(columns={"ride_lengt" : 'ride_length'})
```

```python
"""
Mengecek apakah column 2019 sudah sesuai dengan 2020
"""
print(df2019.columns.tolist())
print(df2020.columns.tolist())
```

```python
"""
Melakukan penggabungan data menjadi 1 dataframe
"""
df = pd.concat([df2019, df2020], ignore_index=True)
```

```python
"""
Mengecek apakah jumlah baris pada dataframe sudah sesuai
"""
len(df) == len(df2019) + len(df2020)
```

```python
"""
Mengubah nama value pada kolom member_casual menjadi member dan casual
"""

df['member_casual'] = df['member_casual'].replace({'Customer':'casual','Subscriber':'member'})


"""
Mengubah tipe data pada kolom started_at dan ended_at menjadi datetime
"""

for column in ['started_at','ended_at'] :
  df[column] = pd.to_datetime(
      df[column],
      format = 'mixed',
      dayfirst = True,
      errors = 'coerce'
  )
```

```python
"""
Mengecek apakah data type sudah sesuai
"""
df[['started_at','ended_at']].dtypes
```

```python
"""
Menambahkan beberapa column untuk memudahkan analisis data
"""

df['ride_length'] = pd.to_timedelta(df['ride_length'])
df['ride_length_second'] = df['ride_length'].dt.total_seconds()
df['ride_length_minute'] = round(df['ride_length_second']/60,1)

df['year'] = df['started_at'].dt.year
df['month'] = df['started_at'].dt.month
df['month_name'] = df['started_at'].dt.month_name()
df["start_hour"] = df["started_at"].dt.hour
```

```python
df.head(10)
```

```python
file_name = "Trip_Q1_2019_vs_Trip_Q1_2020.csv"

df.to_csv(file_name, index=False)

print(f"Berhasil export: {file_name}")
```
